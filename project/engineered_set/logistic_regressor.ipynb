# I will examine the efficacy of a logistic regressor on the engineered dataset to predict team1 win or lose

```python
import pandas as pd
df = pd.read_csv("../../data/rolling_feature_engineered_tier_1_games.csv")
df
```

```python
df.columns
```

I'm going to drop arbitrary data like ids and names

```python
to_drop = []
to_drop.extend([f"team{i}_player{k}_id" for i in range(1,3) for k in range(1,6)])
to_drop.extend([f"team{i}_id" for i in range(1,3)])
to_drop.extend([f"team{i}" for i in range(1,3)])
to_drop.extend(["tournament", "match_id", "game_id", "map_id", "map_name", "datetime"])
df = df.drop(to_drop, axis=1)
df
```

```python
df.columns
```

Note that even after dropping that data, we still have a lot of dimensions to this data (63 features)

Let's now one-hot encode bestOf (I suspect this has little influence on the result whatsoever, but we will see)

```python
bestOf_cats = pd.get_dummies(df.bestOf, drop_first=True, prefix="bestOf", dtype=int)
df = df.drop("bestOf", axis=1)
```

```python
df.head(5)
```

Now, let's do our train/test split and do feature standardization

```python
import math
from sklearn.preprocessing import StandardScaler
split_point = math.ceil(len(df) * 0.8)
# SPLIT MUST BE TEMPORAL TO AVOID TEST LEAKAGE INTO TRAIN SINCE
# FEATURES ARE CUMULATIVE
train = df.iloc[:split_point] # 0 to split_point - 1
test = df.iloc[split_point:] # split_point to len(df)
X_train = train.drop("team1_win", axis=1)
y_train = train.team1_win
X_test = test.drop("team1_win", axis=1)
y_test = test.team1_win
stnd = StandardScaler().set_output(transform="pandas")
X_train = stnd.fit_transform(X_train)
X_test = stnd.transform(X_test)
```

Now add back the categorical features

```python
train_cat = bestOf_cats.iloc[:split_point]
test_cat = bestOf_cats.iloc[split_point:]
X_train = pd.concat([X_train, train_cat], axis=1)
X_test = pd.concat([X_test, test_cat], axis=1)
display(X_train.head())
display(X_test.head())
```

Now we can train our logistic regressor

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
lr.score(X_train, y_train)
```

```python
((y_train == y_train.mode()[0]).sum()) / len(y_train)
```

It is 59% accurate, so it only performs ~5% better than the baseline.

```python
lr.score(X_test, y_test)
```

```python
((y_test == y_train.mode()[0]).sum()) / len(y_test)
```

The logistic regressor performs ~0.4% worse on the test set compared to the baseline.

I'm curious which features it deems most important

```python
display(X_train.columns)
display(lr.coef_)
```

```python
weights = pd.DataFrame()
weights["Feature"] = X_train.columns
weights["Weight"] = lr.coef_[0]
weights = weights.sort_values(by="Weight", ascending = False)
weights.head(10)
```

```python
weights.tail(10)
```

It is giving the highest absolute weight to best of 3.0, player3_kast, team2_previous_10,average_map_score, and so forth. Totally contrary to my guess, the model seems to highly value bestOf. It also seems to have trouble keeping stats influences consistent in the sense that for some reason team 1 player 3 kast is significantly more heavily weighted than the other team 1 players' kast. Player 3 is arbitrary, so it does not make sense that player 3 should have more weight than others for the kast category. I assume this will improve when we aggregate to team-level stats since we are trying to make a team-level prediction. If the players were ordered by performance, such as where player 1 is always the player with the highest stats, then maybe it could give a more fair assessment to the weights, but with the arbitrary ordering of players, it is hard to assign a consistent weight. This ordering issue will be removed with team-level aggregation of stats. Collinearity might be an issue as well and could explain why some stats like team 1 player 2 average adr contributes a negative weight which we wouldn't expect.

I assume if we add the interaction term of how the 5 players stats interact per team, this will be much more predictive and the weight will be more meaningful for a team-level prediction.
