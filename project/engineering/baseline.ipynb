# We begin with classification, so we fit a baseline classifier that predicts the mode

```python
import pandas as pd
data = "../../data/rolling_feature_engineered_tier_1_games.csv"
df = pd.read_csv(data)
df
```

# Baseline accuracy rate

We always do a 80/20 chronological split

```python
import math
y = df.team1_win
split_point = math.ceil(len(y) * 0.8)
y_train = y[:split_point]
y_test = y[split_point:]
```

```python
((y_test == y_train.mode()[0]).sum())/len(y_test)
```

```python
((y_train == y_train.mode()[0]).sum())/len(y_train)
```

Test and train are roughly equally distributed. Baseline is 55.8% accurate on test, 54.6% accurate on train.
