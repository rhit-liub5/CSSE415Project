Given previous findings, we find it plausible that the previous models are modeling noise of player position shuffling since player 1-5 is arbitrary and not sorted by skill. Knowing this, to remove the ordering factor, we will collapse the data into aggregative team-level average statistics, but also add standard deviation and range as well to denote the presence of a carry on the team.

```python
import pandas as pd
df = pd.read_csv("../../data/rolling_feature_engineered_tier_1_games.csv")
df.shape
```

```python
df.columns
```

```python
stats = ["kills", "deaths", "assists", "adr", "kast", "kddiff"]
for i in range(1, 3):
    for stat in stats:
        stat_cols_for_team = [f"previous_10_game_team{i}_player{k}_average_{stat}" for k in range(1, 6)]
        df[f"previous_10_games_team{i}_average_{stat}"] = df[stat_cols_for_team].mean(axis=1)
        df[f"previous_10_games_team{i}_std_{stat}"] = df[stat_cols_for_team].std(ddof=0, axis=1)
        df[f"previous_10_games_team{i}_range_{stat}"] = df[stat_cols_for_team].max(axis=1) - df[stat_cols_for_team].min(axis=1)
        df[f"previous_10_games_team{i}_max_{stat}"] = df[stat_cols_for_team].max(axis=1)
        df[f"previous_10_games_team{i}_min_{stat}"] = df[stat_cols_for_team].min(axis=1)
        df[f"previous_10_games_team{i}_median_{stat}"] = df[stat_cols_for_team].median(axis=1)
        df = df.drop(stat_cols_for_team, axis=1)
```
